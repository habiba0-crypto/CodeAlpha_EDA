```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the uploaded CSV file
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name, sep=None, engine="python")

print("Dataset loaded successfully!")
```

```python

df.head()
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python

df.info()
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Check dataset information
print("\nDataset Description:")
print(df.describe())
```

```python
# Count books for each rating
rating_counts = df['Rating'].value_counts().sort_index()

print("Books by Rating:")
print(rating_counts)
```

```python
plt.figure(figsize=(8, 5))

rating_counts.plot(kind='bar')

plt.title('Distribution of Book Ratings')
plt.xlabel('Rating (Stars)')
plt.ylabel('Number of Books')
plt.xticks(rotation=0)

plt.show()
```

```python
# Price distribution
plt.figure(figsize=(8, 5))

plt.hist(df['Price'], bins=10)

plt.title('Distribution of Book Prices')
plt.xlabel('Book Price')
plt.ylabel('Number of Books')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Rating'], df['Price'])

plt.title('Relationship Between Book Price and Rating')
plt.xlabel('Book Rating')
plt.ylabel('Book Price')

plt.show()
```

```python
# Count books by availability
print(df['Availability'].value_counts())

# Create availability chart
plt.figure(figsize=(6, 4))

df['Availability'].value_counts().plot(kind='bar')

plt.title('Book Availability')
plt.xlabel('Availability')
plt.ylabel('Number of Books')
plt.xticks(rotation=0)

plt.show()
```
